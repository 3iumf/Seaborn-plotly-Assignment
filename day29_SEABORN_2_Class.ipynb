## SEABORN
- Is a python data visualization library built on top of Matplotlib
- Used for Statistical Visualization
- Also used for data exploration
- Works with pandas dataframe
- Creating attractive charts with less code
- imported as sns

```python
import pandas as pd

data = {
    "Name": ["John", "Mary", "Peter", "Sarah", "David", "Jane"],
    "Age": [21, 25, 30, 28, 35, 23],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Department": ["IT", "HR", "IT", "Sales", "Sales", "HR"],
    "Salary": [120000, 150000, 180000, 170000, 220000, 140000]
}

import pandas as pd
df = pd.DataFrame(data)
```

```python
df.head(2)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

#### Scatter Plot

```python
sns.scatterplot(data=df, x="Age", y="Salary", hue="Gender")
plt.show()
```

#### Line Plot

```python
sales = pd.DataFrame({
    "Month" : ['Jan', 'Feb', 'Mar', 'Apr'],
    "Sales": [100, 200, 400, 350]
})
```

```python
sns.lineplot(data=sales, x="Month", y="Sales", marker="o")
```

#### Bar Plot

```python
sns.barplot(data=df, x="Department", y="Salary")
```

#### Count Plot

```python
sns.countplot(data=df, x="Department")
```

#### Histogram

```python
sns.set_theme(style="whitegrid")
sns.histplot(data=df, x="Salary", bins=5)
```

#### Box Plot
- Helps you to understand and the spread of numerical data
- A box plot in Seaborn is used to visualize the distribution of numerical data.
- It shows the median, quartiles, minimum and maximum values of the dataset.
- It is useful for identifying outliers—values that are unusually high or low.
- Box plots make it easy to compare distributions between different categories.

```python
sns.boxplot(data=df, x="Department", y="Salary")
```

```python
data = {
    "Product": ["Laptop", "Phone", "Tablet", "Monitor", "Keyboard", "Mouse"],
    "Category": ["Computer", "Mobile", "Mobile", "Computer", "Accessory", "Accessory"],
    "Price": [800, 600, 350, 300, 80, 40],
    "Sales": [25, 50, 35, 30, 80, 100],
    "Rating": [4.5, 4.7, 4.2, 4.3, 4.6, 4.4]
}

df = pd.DataFrame(data)

```

```python
#A bar chart showing sales by product
sns.set_theme(style="whitegrid")
sns.barplot(data=df, x="Product", y="Sales")
plt.title("Sales by Product")


sns.scatterplot(data=df, x="Product", y="Sales", hue="Category")
plt.title("Price Vs Sales")

sns.histplot(data=df, x="Price", bins=5)
plt.title("Distribution of Product Prices")

sns.boxplot(data=df, x="Category", y="Price")
plt.title("Product Prices by Category")




fig = px.scatter(df, x="Price", y="Sales", hover_name="Product", color="Category")
fig.show()
```

```python
fig = px.scatter(df, x="Price", y="Sales")
fig.show()
```

```python
fig = px.bar(df, x="Product", y="Sales", color="Category", 
             hover_data=["Rating"], hover_name="Product", title="Price Vs Sales")

fig.update_layout(
    xaxis_title= "Product",
    yaxis_title= "Number of Sales"
)
```
